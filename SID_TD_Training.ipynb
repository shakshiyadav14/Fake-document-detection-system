<h3> Get Dataset </h3>

```python
import requests
import os
import zipfile
from tqdm.auto import tqdm

DATA_ROOT = "/content/SIDTD_data"
os.makedirs(DATA_ROOT, exist_ok=True)

TEMPLATE_ZIP = os.path.join(DATA_ROOT, "templates.zip")

urls = [
    "http://datasets.cvc.uab.es/SIDTD/templates.zip",
    "https://datasets.cvc.uab.es/SIDTD/templates.zip"
]

downloaded = False

for url in urls:
    try:
        print("Trying:", url)

        with requests.get(
            url,
            stream=True,
            timeout=60
        ) as response:

            response.raise_for_status()

            total = int(
                response.headers.get("content-length", 0)
            )

            with open(TEMPLATE_ZIP, "wb") as f:
                with tqdm(
                    total=total if total > 0 else None,
                    unit="B",
                    unit_scale=True,
                    desc="Downloading SIDTD templates"
                ) as progress:

                    for chunk in response.iter_content(
                        chunk_size=1024 * 1024
                    ):
                        if chunk:
                            f.write(chunk)
                            progress.update(len(chunk))

        if not zipfile.is_zipfile(TEMPLATE_ZIP):
            raise ValueError(
                "Downloaded file is not a valid ZIP archive."
            )

        downloaded = True
        print("Download completed successfully.")
        break

    except Exception as e:
        print("Download attempt failed:", e)

if not downloaded:
    raise RuntimeError(
        "Automatic download failed. Download templates.zip "
        "from the official SIDTD page and upload it to Colab."
    )

print("Archive:", TEMPLATE_ZIP)
print("Size (GB):", round(os.path.getsize(TEMPLATE_ZIP) / 1e9, 3))
```

<h4> Load Images </h4>

```python
from pathlib import Path
from collections import Counter

# Paths to document images
REAL_DIR = Path("/content/SIDTD/templates/templates/Images/reals")
FAKE_DIR = Path("/content/SIDTD/templates/templates/Images/fakes")

# Supported image formats
IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}

# Check folders
assert REAL_DIR.exists(), f"Missing folder: {REAL_DIR}"
assert FAKE_DIR.exists(), f"Missing folder: {FAKE_DIR}"

# Label convention:
# 0 = genuine
# 1 = forged

samples = []

for path in REAL_DIR.rglob("*"):
    if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS:
        samples.append((str(path), 0))

for path in FAKE_DIR.rglob("*"):
    if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS:
        samples.append((str(path), 1))

print("Total images:", len(samples))
print("Class counts:", Counter(label for _, label in samples))
print("First 10 samples:")

for sample in samples[:10]:
    print(sample)
```

```python
from sklearn.model_selection import train_test_split

# Separate image paths and labels
image_paths = [path for path, label in samples]
labels = [label for path, label in samples]

# 80% training, 20% temporary
train_paths, temp_paths, train_labels, temp_labels = train_test_split(
    image_paths,
    labels,
    test_size=0.20,
    random_state=42,
    stratify=labels
)

# Split remaining 20% equally: 10% validation, 10% test
val_paths, test_paths, val_labels, test_labels = train_test_split(
    temp_paths,
    temp_labels,
    test_size=0.50,
    random_state=42,
    stratify=temp_labels
)

print("Training images:", len(train_paths))
print("Validation images:", len(val_paths))
print("Testing images:", len(test_paths))
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from torchvision import transforms

# Use the device variable from your earlier cells,
# or define it again safely.
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Image transformations
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

eval_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


class SIDTDDataset(Dataset):
    def __init__(self, paths, labels, transform=None):
        self.paths = paths
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, idx):
        image = Image.open(self.paths[idx]).convert("RGB")
        label = self.labels[idx]

        if self.transform:
            image = self.transform(image)

        return image, torch.tensor(label, dtype=torch.long)


# Create datasets
train_dataset = SIDTDDataset(
    train_paths, train_labels, train_transform
)

val_dataset = SIDTDDataset(
    val_paths, val_labels, eval_transform
)

test_dataset = SIDTDDataset(
    test_paths, test_labels, eval_transform
)

# Create data loaders
BATCH_SIZE = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=(device.type == "cuda")
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=(device.type == "cuda")
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=(device.type == "cuda")
)

print("DataLoaders created successfully!")
print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Testing batches:", len(test_loader))
```
