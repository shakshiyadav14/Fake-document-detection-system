```python
import os
import random
from faker import Faker
from PIL import Image, ImageDraw, ImageFilter, ImageFont

# # 1. Setup Directories
# os.makedirs("data/templates", exist_ok=True)
# os.makedirs("data/dataset/genuine", exist_ok=True)
# os.makedirs("data/dataset/tampered", exist_ok=True)
# os.makedirs("data/faces", exist_ok=True)

fake = Faker("en_IN")


# Helper: Generate realistic wood texture background
def create_wood_background(width=1024, height=576):
    bg = Image.new("RGB", (width, height), color=(222, 184, 135))
    draw = ImageDraw.Draw(bg)

    for _ in range(400):
        y = random.randint(0, height)
        shade = random.randint(190, 235)
        draw.line(
            [(0, y), (width, y + random.randint(-15, 15))],
            fill=(shade, shade - 30, shade - 60),
            width=random.randint(1, 3),
        )

    shadow = Image.new("RGBA", (width, height), (0, 0, 0, 0))
    s_draw = ImageDraw.Draw(shadow)
    s_draw.polygon(
        [(width // 2, 0), (width, 0), (width, height), (width // 3, height)],
        fill=(0, 0, 0, 45),
    )
    shadow = shadow.filter(ImageFilter.GaussianBlur(50))

    bg = bg.convert("RGBA")
    bg.alpha_composite(shadow)
    return bg.convert("RGB")


def draw_rounded_rectangle(
    draw, bounds, radius, fill=None, outline=None, width=1
):
    x1, y1, x2, y2 = bounds
    draw.rounded_rectangle(
        [x1, y1, x2, y2],
        radius=radius,
        fill=fill,
        outline=outline,
        width=width,
    )


# ---------------------------------------------------------
# STEP 1: CREATE BASE TEMPLATE
# ---------------------------------------------------------
def create_specimen_template():
    template_path = "data/templates/specimen_template.png"

    bg = create_wood_background(1024, 576)
    card_x1, card_y1, card_x2, card_y2 = 140, 160, 510, 410

    # Soft Drop Shadow
    shadow_layer = Image.new("RGBA", bg.size, (0, 0, 0, 0))
    sh_draw = ImageDraw.Draw(shadow_layer)
    draw_rounded_rectangle(
        sh_draw,
        (card_x1 + 6, card_y1 + 8, card_x2 + 6, card_y2 + 8),
        radius=12,
        fill=(0, 0, 0, 90),
    )
    shadow_layer = shadow_layer.filter(ImageFilter.GaussianBlur(8))
    bg = Image.alpha_composite(bg.convert("RGBA"), shadow_layer).convert("RGB")

    draw = ImageDraw.Draw(bg)
    draw_rounded_rectangle(
        draw,
        (card_x1, card_y1, card_x2, card_y2),
        radius=12,
        fill=(245, 248, 252),
        outline=(120, 150, 180),
        width=2,
    )
    draw_rounded_rectangle(
        draw,
        (card_x1 + 6, card_y1 + 6, card_x2 - 6, card_y2 - 6),
        radius=10,
        fill=None,
        outline=(0, 51, 102),
        width=2,
    )

    try:
        font_header = ImageFont.truetype("arialbd.ttf", 12)
        font_sub = ImageFont.truetype("arial.ttf", 10)
        font_label = ImageFont.truetype("arialbd.ttf", 11)
    except:
        font_header = font_sub = font_label = ImageFont.load_default()

    draw.text(
        (card_x1 + 18, card_y1 + 15),
        "'INSTITUTE OF DIGITAL IDENTITY & INNOVATION",
        fill=(0, 0, 0),
        font=font_header,
    )
    draw.text(
        (card_x1 + 45, card_y1 + 30),
        "SPECIMEN CARD / EDUCATIONAL LAYOUT",
        fill=(30, 30, 30),
        font=font_sub,
    )
    draw.ellipse(
        (card_x2 - 42, card_y1 + 12, card_x2 - 18, card_y1 + 36),
        fill=(218, 165, 32),
        outline=(184, 134, 11),
    )

    draw.rectangle(
        [
            (card_x1 + 13),
            (card_y1 + 55),
            (card_x1 + 120),
            (card_y1 + 170),
        ],
        fill=(220, 220, 220),
        outline=(100, 100, 100),
        width=1,
    )
    draw.text(
        (card_x1 + 30, card_y1 + 173),
        "Profile Photo",
        fill=(50, 50, 50),
        font=font_sub,
    )

    labels_y = card_y1 + 56
    draw.text(
        (card_x1 + 130, labels_y), "Name:", fill=(0, 0, 0), font=font_label
    )
    draw.text(
        (card_x1 + 130, labels_y + 26),
        "Father's Name:",
        fill=(0, 0, 0),
        font=font_label,
    )
    draw.text(
        (card_x1 + 130, labels_y + 44),
        "Date of Birth:",
        fill=(0, 0, 0),
        font=font_label,
    )
    draw.text(
        (card_x1 + 130, labels_y + 62),
        "Gender:",
        fill=(0, 0, 0),
        font=font_label,
    )
    draw.text(
        (card_x1 + 130, labels_y + 80),
        "Aadhar Number:",
        fill=(0, 0, 0),
        font=font_label,
    )
    draw.text(
        (card_x1 + 130, labels_y + 98),
        "Virtual ID:",
        fill=(0, 0, 0),
        font=font_label,
    )
    draw.text(
        (card_x1 + 130, labels_y + 116),
        "Issue Date:",
        fill=(0, 0, 0),
        font=font_label,
    )

    bg.save(template_path, "PNG")


create_specimen_template()


# ---------------------------------------------------------
# STEP 2: MULTI-FIELD TAMPERING DATASET GENERATOR
# ---------------------------------------------------------
def generate_specimen_card(card_id, is_tampered=False):
    template_path = "data/templates/specimen_template.png"
    card = Image.open(template_path).convert("RGB")
    draw = ImageDraw.Draw(card)

    card_x1, card_y1 = 140, 160

    try:
        font_bold = ImageFont.truetype("arialbd.ttf", 12)
        font_val = ImageFont.truetype("arial.ttf", 11)
    except:
        font_bold = font_val = ImageFont.load_default()

    # Generate Authentic Base Values
    name = f"{fake.first_name()[0]}.{fake.first_name()[0]}. {fake.last_name().upper()}"
    father_name = f"K.R. {fake.last_name().upper()}"
    dob = fake.date_of_birth(minimum_age=18, maximum_age=60).strftime(
        "%d / %m / %Y"
    )
    gender = random.choice(["MALE", "FEMALE"])
    uid = f"{random.randint(1000,9999)} {random.randint(1000,9999)} {random.randint(1000,9999)}"
    vid = f"{random.randint(1000,9999)} {random.randint(1000,9999)} {random.randint(1000,9999)} {random.randint(1000,9999)}"
    issue_date = "01 / 01 / 2026"

    face_files = [
        f for f in os.listdir("data/faces") if f.endswith((".jpg", ".png"))
    ]

    # Handle Genuine vs. Tampered Logic
    tampered_fields = []

    if is_tampered:
        # Pick 1, 2, or 3 fields to tamper simultaneously
        possible_types = [
            "photo_swap",
            "fake_name",
            "fake_dob",
            "fake_uid",
            "qr_mismatch",
        ]
        num_tampers = random.randint(1, 3)
        tampered_fields = random.sample(possible_types, k=num_tampers)

        # Apply Field Modifications
        if "fake_name" in tampered_fields:
            name = (
                f"S.K. {fake.last_name().upper()}"  # Modifies text value on card
            )

        if "fake_dob" in tampered_fields:
            dob = "10 / 10 / 1999"  # Alters birth date string

        if "fake_uid" in tampered_fields:
            uid = "9999 0000 1111"  # Alters 12-digit UID string

    # 1. Paste Primary Face
    if face_files:
        chosen_face = os.path.join("data/faces", random.choice(face_files))
        face_img = Image.open(chosen_face).convert("RGB").resize((107, 115))
        card.paste(face_img, (card_x1 + 13, card_y1 + 55))

        # Apply Photo Tampering (if selected)
        if "photo_swap" in tampered_fields and len(face_files) > 1:
            intruder = os.path.join("data/faces", random.choice(face_files))
            intruder_img = (
                Image.open(intruder).convert("RGB").resize((103, 110))
            )
            card.paste(
                intruder_img, (card_x1 + 15, card_y1 + 57)
            )  # Spliced/Misaligned Paste

    # 2. Render Text Values onto Template
    val_x = card_x1 + 225
    labels_y = card_y1 + 56

    draw.text((val_x - 40, labels_y), name, fill=(0, 0, 0), font=font_bold)
    draw.text(
        (val_x + 5, labels_y + 26), father_name, fill=(0, 0, 0), font=font_val
    )
    draw.text((val_x + 5, labels_y + 44), dob, fill=(0, 0, 0), font=font_val)
    draw.text((val_x + 5, labels_y + 62), gender, fill=(0, 0, 0), font=font_val)
    draw.text((val_x + 5, labels_y + 80), uid, fill=(0, 0, 0), font=font_val)
    draw.text((val_x + 5, labels_y + 98), vid, fill=(0, 0, 0), font=font_val)
    draw.text(
        (val_x + 5, labels_y + 116), issue_date, fill=(0, 0, 0), font=font_val
    )

    # Save Generated Sample
    folder = "tampered" if is_tampered else "genuine"
    save_path = f"data/dataset/{folder}/card_{card_id}.jpg"
    card.save(save_path, "JPEG", quality=92)

    if is_tampered:
        print(f"Generated Tampered Card #{card_id}: {tampered_fields}")
    else:
        print(f"Generated Genuine Card #{card_id}")


# Run Batch Generation (50 Genuine, 50 Tampered)
for i in range(5):
    generate_specimen_card(i, is_tampered=False)
    generate_specimen_card(i, is_tampered=True)
```
